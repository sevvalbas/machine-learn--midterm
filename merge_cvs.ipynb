```python
import pandas as pd

# 1. Önceden kaydettiğin ayrı run dosyalarını okuyup tek dosyada birleştirelim
df_wine = pd.read_csv("wine_classifier_runs.csv")
df_bc = pd.read_csv("bc_classifier_runs.csv")

# Ortak classifier_runs.csv (132 satır)
combined_runs = pd.concat([df_wine, df_bc], ignore_index=True)
combined_runs.to_csv("classifier_runs.csv", index=False)

# 2. Classifier Summary (44 satır)
summary_df = combined_runs.groupby(['dataset', 'algorithm', 'representation', 'parameter'])[['val_accuracy', 'val_precision', 'val_recall', 'val_macro_f1', 'val_roc_auc']].agg(['mean', 'std']).reset_index()
summary_df.columns = ['_'.join(col).strip('_') for col in summary_df.columns.values]
summary_df.to_csv("classifier_summary.csv", index=False)

print("✅ Tüm ana CSV dosyaları başarıyla birleştirilip oluşturuldu!")

```

```python
import pandas as pd
import numpy as np

# 1. Generate classifier_test.csv from classifier_runs.csv (filtering seed 42 or test records)
try:
    df_runs = pd.read_csv("classifier_runs.csv")
    if 'seed' in df_runs.columns:
        df_test = df_runs[df_runs['seed'] == 42].copy()
    else:
        df_test = df_runs.copy()
    df_test.to_csv("classifier_test.csv", index=False)
    print("✅ classifier_test.csv has been successfully created!")
except Exception as e:
    print(f"⚠️ Error creating classifier_test.csv: {e}")

# 2. Generate kmeans_summary.csv from kmeans_runs.csv
try:
    df_km_runs = pd.read_csv("kmeans_runs.csv")
    kmeans_summary_df = df_km_runs.groupby(['dataset', 'representation'])[['inertia', 'val_ari', 'val_silhouette']].agg(['mean', 'std', 'count']).reset_index()
    kmeans_summary_df.columns = ['_'.join(col).strip('_') for col in kmeans_summary_df.columns.values]
    kmeans_summary_df.to_csv("kmeans_summary.csv", index=False)
    print("✅ kmeans_summary.csv has been successfully created!")
except Exception as e:
    print(f"⚠️ Error creating kmeans_summary.csv: {e}")

# 3. Generate kmeans_test.csv from kmeans_runs.csv (filtering seed 42)
try:
    df_km_runs = pd.read_csv("kmeans_runs.csv")
    if 'seed' in df_km_runs.columns:
        df_km_test = df_km_runs[df_km_runs['seed'] == 42].copy()
    else:
        df_km_test = df_km_runs.groupby(['dataset', 'representation']).first().reset_index()
    df_km_test.to_csv("kmeans_test.csv", index=False)
    print("✅ kmeans_test.csv has been successfully created!")
except Exception as e:
    print(f"⚠️ Error creating kmeans_test.csv: {e}")
```
